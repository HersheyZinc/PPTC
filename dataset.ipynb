```python
import json, glob
from tqdm import tqdm
from src.utils import read_lines
from pptx import Presentation
from src.ppt_reader import get_content
```

```python
filepaths = glob.glob('PPT_test_input/Edit_PPT_template/*.json')
```

```python
# import random
# sampled_filepaths = random.sample(filepaths, len(filepaths) // 2)
for filepath in tqdm(filepaths):
    lines = read_lines(filepath)
    session_no = filepath.split("_")[-1].split(".")[0]
    messages = []
    for line in lines:
        turn = json.loads(line)
        turn_no = turn["Turn"]
        with open(f"PPT_Prompt_File/Edit_PPT_template/tf_gpt4o-mini_n4_{session_no}_{turn_no}.txt") as f:
            prompt = f.read()
            messages.append({"role":"user", "content":"All the PPT contents are:\n<Begin of PPT>"+ prompt.split("<Begin of PPT>")[-1]})
        messages.append({"role":"assistant", "content": "<code>\n"+turn["Feasible API sequence"]+ "\n</code>"})

    with open('train.jsonl', 'a', encoding='utf-8') as f:
        json.dump({"messages": messages}, f)
        f.write('\n')
```

```python
filepaths = glob.glob('PPT_test_input/Create_new_slides/*.json')
for filepath in tqdm(filepaths):
    lines = read_lines(filepath)
    session_no = filepath.split("_")[-1].split(".")[0]
    messages = []
    for line in lines:
        turn = json.loads(line)
        turn_no = turn["Turn"]
        with open(f"PPT_Prompt_File/Create_new_slides/tf_gpt4o-mini_n4_{session_no}_{turn_no}.txt") as f:
            prompt = f.read()

            messages.append({"role":"user", "content":"All the PPT contents are:\n<Begin of PPT>"+ prompt.split("<Begin of PPT>")[-1]})
        messages.append({"role":"assistant", "content": "<code>\n"+turn["Feasible API sequence"]+ "\n</code>"})

    
    with open('eval.jsonl', 'a', encoding='utf-8') as f:
        json.dump({"messages": messages}, f)
        f.write('\n')
```
